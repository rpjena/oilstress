```python
import pandas as pd
import numpy as np
import json
```

```python
# Case 3: From a .json file
with open("scenario_shocks.json") as f:
    data = json.load(f)
df = pd.DataFrame(data)
```

```python
shocks_data = df.loc['GLOBAL_BLOWUP'].scenarios['shocks']
# Mean each list, then transpose so assets=rows, scenarios=columns
shocks_df = pd.DataFrame(shocks_data).T                          # shape: (4 assets × 3 scenarios) after .T
shocks_df = shocks_df.map(np.mean)                               # collapse each [lo, hi] → scalar

print(shocks_df)
```
